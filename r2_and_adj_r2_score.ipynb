# CASE-1:

 **The data only with yearexperience and salary**

```python
import numpy as np
import pandas as pd
import sklearn
```

```python
df=pd.read_csv('/content/case_1.csv')
df
```

```python
X=df[['YearsExperience']] # independent

y=df['Salary'] #dependent
```

```python
from sklearn.linear_model import LinearRegression
reg=LinearRegression()
reg.fit(X,y)
```

```python
ans_of_model=reg.predict(X)
```

```python
from sklearn.metrics import root_mean_squared_error
root_mean_squared_error(y,ans_of_model)
```

```python
from sklearn.metrics import r2_score

print(f'The Accuracy {r2_score(y,ans_of_model)}') #r2_score
```

```python
X.shape[1]
```

```python
#adjust_r2_score


value = 1 - ((1-r2_score(y,ans_of_model)) * (len(df)-1)) / ((len(df)-(X.shape[1])-1))

print(f'The Accuracy {value}')
```

# CASE - 2

**The data with yearsexperience and salary with one add coloumn of height**

checking wheather the column add is usefull or not

```python
import numpy as np
import pandas as pd
import sklearn
```

```python
df=pd.read_csv('/content/case_2.csv')
df.head()
```

```python
X=df[['YearsExperience','Height']] # independent

y=df['Salary'] #dependent
```

```python
from sklearn.linear_model import LinearRegression
reg=LinearRegression()
reg.fit(X,y)
```

```python
ans_of_model=reg.predict(X)
```

```python
from sklearn.metrics import root_mean_squared_error
root_mean_squared_error(y,ans_of_model)
```

```python
from sklearn.metrics import r2_score

print(f'The Accuracy {r2_score(y,ans_of_model)}') #r2_score
```

```python
X.shape[1]
```

```python
#adjust_r2_score


value = 1 - ((1-r2_score(y,ans_of_model)) * (len(df)-1)) / ((len(df)-(X.shape[1])-1))

print(f'The Accuracy {value}')
```
